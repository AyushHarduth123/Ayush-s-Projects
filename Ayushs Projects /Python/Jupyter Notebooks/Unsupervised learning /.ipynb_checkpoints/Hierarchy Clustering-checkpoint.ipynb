```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler


df_new = pd.read_excel('Mall Customers.xlsx')
```

```python
x = df_new[["Annual Income (k$)", "Spending Score (1-100)"]]
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram,linkage
```

```python
# Determines the distance between the clusters, distance, number of points 
z = linkage(x_scaled,method='ward')


# the shorter the vertical lines are the more seperated the clusters are 
plt.figure(figsize = (10,5))
dendrogram(z)
plt.title("Dendrogram")
plt.xlabel("clusters")
plt.ylabel("Distance between the clusters ")
plt.show()
```

```python
model=AgglomerativeClustering(n_clusters=5,linkage = 'ward')
df_new['Cluster'] = model.fit_predict(x_scaled)
```
