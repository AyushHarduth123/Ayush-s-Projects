```python
import numpy as np

# Sigmoid activation function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))  
# -------------------------
# Input layer
# -------------------------
x1 = 0.1

# -------------------------
# Hidden layer parameters
# -------------------------
w1 = 0.15
b1 = 0.40

# Hidden layer computation
z1 = w1 * x1 + b1
a1 = sigmoid(z1)
# -------------------------
# Output layer parameters
# -------------------------
w2 = 0.45
b2 = 0.65
# Output layer computation
z2 = w2 * a1 + b2
a2 = sigmoid(z2)
  
# Display results
print("Input x1:", x1)
print("\n--- Hidden Layer ---")
print("z1 = w1 * x1 + b1 =", z1)
print("a1 = sigmoid(z1) =", a1)

print("\n--- Output Layer ---")
print("z2 = w2 * a1 + b2 =", z2)
print("a2 = sigmoid(z2) =", a2)
```
