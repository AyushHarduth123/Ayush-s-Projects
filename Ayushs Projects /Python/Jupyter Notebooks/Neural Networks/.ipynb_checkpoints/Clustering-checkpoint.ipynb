```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data = {
    'Age': [22,25,26,45,46,48,30,32,34,50],
    'AnnualSpend': [15,17,18,60,62,65,25,28,30,70]
}
df = pd.DataFrame(data)
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(df)
```

```python
plt.scatter(df['Age'], df['AnnualSpend'], c=df['Cluster'])
plt.xlabel('Age')
plt.ylabel('Annual Spend (R000)')
plt.title('Customer Clusters')
plt.show()
```
