```python
# -------------------------------
# 1. Import libraries
# -------------------------------
import numpy as np

# -------------------------------
# 2. Dataset
# -------------------------------
# Features: [Glucose, BMI, Age]
X = np.array([
    [85, 22, 25],
    [90, 24, 30],
    [120, 35, 45],
    [140, 40, 50],
    [130, 38, 48],
    [95, 26, 33],
    [110, 30, 40],
    [150, 42, 55]
])

# Labels: 0 = No Diabetes, 1 = Diabetes
y = np.array([0, 0, 1, 1, 1, 0, 1, 1])

# -------------------------------
# 3. Feature scaling
# -------------------------------
X_mean = X.mean(axis=0)
X_std = X.std(axis=0)
X_scaled = (X - X_mean) / X_std

# -------------------------------
# 4. Activation functions
# -------------------------------
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(a):
    return a * (1 - a)

# -------------------------------
# 5. Initialize weights and biases
# -------------------------------
np.random.seed(1)

W1 = np.random.randn(3, 4)   # Input → Hidden
b1 = np.zeros((1, 4))

W2 = np.random.randn(4, 1)   # Hidden → Output
b2 = np.zeros((1, 1))

learning_rate = 0.1
epochs = 1

# -------------------------------
# 6. Train the neural network
# -------------------------------
for epoch in range(epochs):

    # Forward propagation
    z1 = np.dot(X_scaled, W1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, W2) + b2
    y_hat = sigmoid(z2)

    # Loss (Binary Cross-Entropy)
    loss = -np.mean(
        y.reshape(-1, 1) * np.log(y_hat) +
        (1 - y.reshape(-1, 1)) * np.log(1 - y_hat)
    )

    # Backpropagation
    error_output = y_hat - y.reshape(-1, 1)
    dW2 = np.dot(a1.T, error_output)
    db2 = np.sum(error_output)

    error_hidden = np.dot(error_output, W2.T) * sigmoid_derivative(a1)
    dW1 = np.dot(X_scaled.T, error_hidden)
    db1 = np.sum(error_hidden)

    # Update weights
    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2
    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

    # Show progress
    if epoch % 200 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")

# -------------------------------
# 7. Model Evaluation
# -------------------------------
predictions = (y_hat >= 0.5).astype(int)
accuracy = np.mean(predictions.flatten() == y)

print("\nTraining Accuracy:", accuracy)

# -------------------------------
# 8. Predict for a NEW patient
# -------------------------------

# New patient data: [Glucose, BMI, Age]
new_patient = np.array([[125, 33, 26]])

# Apply SAME scaling as training data
new_patient_scaled = (new_patient - X_mean) / X_std

# Forward pass for new data
z1_new = np.dot(new_patient_scaled, W1) + b1
a1_new = sigmoid(z1_new)

z2_new = np.dot(a1_new, W2) + b2
probability = sigmoid(z2_new)

prediction = (probability >= 0.5).astype(int)

print("\nNew Patient Prediction")
print("Probability of Diabetes:", probability[0][0])
print("Prediction (1=Diabetes, 0=No Diabetes):", prediction[0][0])
```
