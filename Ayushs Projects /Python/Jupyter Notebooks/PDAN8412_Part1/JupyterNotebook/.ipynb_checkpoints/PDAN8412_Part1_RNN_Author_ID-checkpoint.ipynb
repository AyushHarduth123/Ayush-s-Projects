# PDAN8412 POE Part 1: Guessing an Author from Writing Style with Recurrent Neural Networks

**Scenario.** I am a data analyst at a large book press. The press wants a fun bot for fans that reads a short passage and guesses which author wrote it, using only the writing style. The bot will be built with a Long Short-Term Memory (LSTM) network, which is a type of recurrent neural network.

**Dataset.** I use the Spooky Author Identification dataset from Kaggle (Kaggle, 2017). The file `train.csv` holds 19 579 sentences, each labelled with one of three authors: Edgar Allan Poe (EAP), Mary Shelley (MWS) and H.P. Lovecraft (HPL). The file `test.csv` has no labels, so it cannot be used to measure accuracy. I therefore split `train.csv` into training, validation and test sets myself.

**How this notebook works.** Spark (PySpark) is used for the data analysis, which covers loading, counting, checking quality and summarising. TensorFlow and Keras are used to train the models, because training does not need to run in Spark. Every code cell is followed by a markdown cell that explains the code in plain English. Each chart is saved as a numbered PNG in the `figures` folder so that it can be reused in the report.

**AI assistance declaration.** Parts of the code in this notebook were drafted with the help of Claude, an AI assistant made by Anthropic (Anthropic, 2026). Code cells that were drafted this way carry an attribution comment at the top. The full references are listed at the end of the notebook.

## 1. Setup and imports

This section prepares everything the rest of the notebook depends on. It loads the libraries, fixes the random seeds so that results can be repeated, sets up the folder paths, chooses a consistent look for the charts, and starts the Spark session.

```python
# Attribution: drafted with the help of Claude (Anthropic, 2026); see the AI assistance declaration above.
import os
import sys
import random
from pathlib import Path

os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"   # keep TensorFlow's start-up messages short

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
import pyspark
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

print("Python    :", sys.version.split()[0])
print("NumPy     :", np.__version__)
print("pandas    :", pd.__version__)
print("PySpark   :", pyspark.__version__)
print("TensorFlow:", tf.__version__)
print("Devices   :", [d.device_type for d in tf.config.list_physical_devices()])
```

This code brings in the libraries used in the first part of the analysis. Pandas and NumPy handle tables and numbers, Matplotlib and Seaborn draw charts, PySpark handles the large-data style analysis, and TensorFlow will be used later to build the neural networks. The last few lines print the version of each library and list the processors that TensorFlow can see, which makes it clear exactly what setup produced the results. Libraries that are only needed later, such as scikit-learn, NLTK and the word cloud tool, are imported in the section where they are first used, so each import sits next to the code that needs it.

```python
# Attribution: drafted with the help of Claude (Anthropic, 2026).
# Fix the random seeds so that every run gives the same splits and the same starting weights.
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# The notebook lives in JupyterNotebook/, so the project folder is one level up.
PROJECT_DIR = Path.cwd().parent
DATA_PATH = PROJECT_DIR / "Traning Data" / "train.csv"   # the folder is spelt "Traning" on disk
FIG_DIR = PROJECT_DIR / "figures"
OUT_DIR = PROJECT_DIR / "outputs"

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Could not find {DATA_PATH}. Open this notebook from the JupyterNotebook folder "
        "so that the project folder sits one level above it."
    )

FIG_DIR.mkdir(exist_ok=True)
OUT_DIR.mkdir(exist_ok=True)

print("Seed        :", SEED)
print("Project     :", PROJECT_DIR.name)
print("Data file   :", DATA_PATH.relative_to(PROJECT_DIR), f"({DATA_PATH.stat().st_size / 1e6:.1f} MB)")
print("Figures dir :", FIG_DIR.relative_to(PROJECT_DIR))
print("Outputs dir :", OUT_DIR.relative_to(PROJECT_DIR))
```

This code sets the random seed to 42 for Python, NumPy and TensorFlow. Many steps in this project involve chance, such as shuffling the data and choosing the starting weights of a network, so fixing the seed means that anyone who runs the notebook again gets the same results. The code then works out where the data, the charts and the model results are stored, using the fact that the notebook sits inside the `JupyterNotebook` folder. It stops with a clear message if the data file cannot be found, and it creates the `figures` and `outputs` folders if they do not exist yet. The data folder is spelt "Traning Data" on disk, so the path uses that exact spelling.

```python
# Attribution: drafted with the help of Claude (Anthropic, 2026).
sns.set_theme(style="whitegrid", context="notebook")

AUTHOR_ORDER = ["EAP", "MWS", "HPL"]
AUTHOR_NAMES = {
    "EAP": "Edgar Allan Poe",
    "MWS": "Mary Shelley",
    "HPL": "H.P. Lovecraft",
}
# One colour per author, taken from a colour-blind friendly palette, reused in every chart.
AUTHOR_COLOURS = {"EAP": "#0072B2", "MWS": "#009E73", "HPL": "#D55E00"}


def save_figure(filename):
    """Save the current chart into figures/ at 150 dpi, then show it."""
    plt.savefig(FIG_DIR / filename, dpi=150, bbox_inches="tight")
    plt.show()
    print(f"Saved: figures/{filename}")


print({code_: f"{AUTHOR_NAMES[code_]} ({AUTHOR_COLOURS[code_]})" for code_ in AUTHOR_ORDER})
```

This code chooses one clean chart style for the whole notebook and sets up the three authors in a fixed order, with their full names and one fixed colour each. Using the same colour for the same author in every chart makes the charts much easier to read and compare. The colours come from a palette that stays distinguishable for readers with colour blindness. The `save_figure` function is a small shortcut that saves the current chart as a PNG at 150 dpi inside the `figures` folder and then shows it, so that every chart used in the report is stored on disk automatically.

```python
# Attribution: drafted with the help of Claude (Anthropic, 2026).
os.environ["SPARK_LOCAL_IP"] = "127.0.0.1"          # keep Spark on this computer; avoids a hostname warning

spark = (
    SparkSession.builder
    .master("local[*]")                              # run on this computer, using all its cores
    .appName("PDAN8412_Part1_Author_Identification")
    .config("spark.driver.memory", "2g")
    .config("spark.sql.shuffle.partitions", "8")     # the data is small, so the default of 200 is wasteful
    .config("spark.ui.showConsoleProgress", "false")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")

print("Spark version      :", spark.version)
print("Master             :", spark.sparkContext.master)
print("Default parallelism:", spark.sparkContext.defaultParallelism)
```

This code starts a local Spark session, which is the entry point for all the Spark work in this notebook. "Local" means Spark runs on this laptop and splits the work across its processor cores instead of using a cluster of computers. A few settings are worth noting. The number of shuffle partitions is lowered from Spark's default of 200 to 8, because the dataset is only about three megabytes and 200 tiny pieces would just slow things down. Spark is told to stay on the local machine, which avoids network warnings and failures on some home and university networks. Log messages are limited to errors so that the notebook output stays readable. The printed lines confirm which Spark version is running and how many cores it can use.
