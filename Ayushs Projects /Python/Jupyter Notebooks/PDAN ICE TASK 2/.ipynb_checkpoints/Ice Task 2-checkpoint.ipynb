# POE Part 2 — Comparative Analysis of Classification Algorithms
**Dataset:** Iris Plants Dataset (sklearn)  
**Task:** Multiclass classification — predict iris species (Setosa, Versicolour, Virginica)  
**Models:** Logistic Regression · Decision Tree · Random Forest · SVC · KNN

## 1. Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

np.random.seed(42)
```

## 2. Load and Prepare Dataset

The **Iris dataset** has 150 samples, 4 features, and 3 balanced classes (50 each):  
- Class 0: Setosa  
- Class 1: Versicolour  
- Class 2: Virginica

```python
iris = load_iris()
x, y = iris.data, iris.target
feature_names = iris.feature_names
target_names  = iris.target_names

df = pd.DataFrame(x, columns=feature_names)
df['Species'] = [target_names[i] for i in y]

print(f'Shape: {df.shape}')
print(f'Features: {list(feature_names)}')
print(f'Classes: {list(target_names)}')
print(f'Missing values: {df.isnull().sum().sum()}')
print()
print('Class distribution:')
print(df['Species'].value_counts())
df.head()
```

```python
# 80/20 train-test split, stratified to keep class balance equal in both sets
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)

# Scale features — required for SVC and KNN which are distance-based
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled  = scaler.transform(x_test)

print(f'Training samples : {len(x_train)}')
print(f'Testing samples  : {len(x_test)}')
print(f'Training class distribution : {np.bincount(y_train)}')
print(f'Testing class distribution  : {np.bincount(y_test)}')
```

## 3. Build Classification Models

```python
classifiers = {
    'Logistic Regression': LogisticRegression(solver='lbfgs', max_iter=1000, random_state=42),
    'Decision Tree'      : DecisionTreeClassifier(max_depth=5, random_state=42),
    'Random Forest'      : RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42),
    'SVC'                : SVC(kernel='rbf', probability=True, random_state=42),
    'KNN'                : KNeighborsClassifier(n_neighbors=5, weights='distance')
}
```

## 4. Evaluate Each Model — Accuracy, Precision, Recall, F1-Score

```python
results = []
predictions = {}

for name, clf in classifiers.items():
    clf.fit(x_train_scaled, y_train)
    y_pred = clf.predict(x_test_scaled)
    predictions[name] = y_pred

    acc  = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, average='weighted', zero_division=0)
    rec  = recall_score(y_test, y_pred, average='weighted', zero_division=0)
    f1   = f1_score(y_test, y_pred, average='weighted', zero_division=0)

    results.append({'Model': name, 'Accuracy': acc, 'Precision': prec, 'Recall': rec, 'F1-Score': f1})

results_df = pd.DataFrame(results).sort_values('Accuracy', ascending=False).reset_index(drop=True)
results_df.index += 1

display_df = results_df.copy()
for col in ['Accuracy','Precision','Recall','F1-Score']:
    display_df[col] = display_df[col].map('{:.2%}'.format)

print('=== Model Performance on Test Set ===')
display_df
```

## 5. Classification Reports (per-class breakdown)

```python
for name, y_pred in predictions.items():
    print('=' * 52)
    print(f'  {name}')
    print('=' * 52)
    print(classification_report(y_test, y_pred, target_names=target_names))
```

## 6. Confusion Matrices

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes = axes.flatten()

for i, (name, y_pred) in enumerate(predictions.items()):
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=target_names)
    disp.plot(ax=axes[i], colorbar=False, cmap='Blues')
    axes[i].set_title(f'{name}\nAccuracy: {accuracy_score(y_test, y_pred):.2%}', fontsize=10)

axes[-1].set_visible(False)
plt.suptitle('Confusion Matrices — All Models', fontsize=13)
plt.tight_layout()
plt.show()
```

## 7. Model Comparison — Bar Chart

```python
metrics = ['Accuracy', 'Precision', 'Recall', 'F1-Score']
x_pos   = np.arange(len(results_df))
width   = 0.18
colors  = ['#2196F3', '#4CAF50', '#FF9800', '#E91E63']

fig, ax = plt.subplots(figsize=(13, 6))

for i, (metric, color) in enumerate(zip(metrics, colors)):
    vals = results_df[metric].values
    bars = ax.bar(x_pos + i * width, vals, width, label=metric, color=color, alpha=0.85)
    for bar, val in zip(bars, vals):
        ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.002,
                f'{val:.2%}', ha='center', va='bottom', fontsize=7.5)

ax.set_xticks(x_pos + width * 1.5)
ax.set_xticklabels(results_df['Model'], rotation=15, ha='right')
ax.set_ylim(0.85, 1.03)
ax.set_ylabel('Score')
ax.set_title('Classification Performance — All Metrics')
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

## 8. Interpretation of Results

### Question 1 — Which model achieved the highest accuracy?

```python
best_acc = results_df.loc[results_df['Accuracy'].idxmax()]
tied     = results_df[results_df['Accuracy'] == best_acc['Accuracy']]['Model'].tolist()
print(f'Highest Accuracy: {best_acc["Accuracy"]:.2%}')
print(f'Model(s): {", ".join(tied)}')
```

**Answer:**  
**SVC** and **KNN** achieved the highest accuracy at **96.67%**, correctly classifying 29 out of 30 test samples. Logistic Regression, Decision Tree, and Random Forest follow at 93.33% (28/30 correct).

SVC achieves this as a single model by finding the optimal margin hyperplane between classes — no ensemble required.

### Question 2 — Which model produced the best balance between precision and recall?

```python
results_df['Prec-Rec Gap'] = abs(results_df['Precision'] - results_df['Recall'])
print(results_df[['Model','Precision','Recall','Prec-Rec Gap']]
      .sort_values('Prec-Rec Gap').to_string(index=False))
```

**Answer:**  
**SVC** produces the best balance between precision and recall with an equal score across all classes, resulting in a near-zero gap between precision and recall.  

This means SVC is equally good at avoiding false positives (precision) and false negatives (recall), which is ideal when no particular type of error is more costly than the other.

### Question 3 — Which model produced the highest F1-Score?

```python
best_f1  = results_df.loc[results_df['F1-Score'].idxmax()]
tied_f1  = results_df[results_df['F1-Score'] == best_f1['F1-Score']]['Model'].tolist()
print(f'Highest F1-Score: {best_f1["F1-Score"]:.4f}')
print(f'Model(s): {", ".join(tied_f1)}')
print()
print(results_df[['Model','F1-Score']].sort_values('F1-Score', ascending=False).to_string(index=False))
```

**Answer:**  
**SVC** and **KNN** share the highest weighted F1-Score at approximately **0.9666**.  

Because the Iris test set is perfectly balanced (10 samples per class), the weighted F1 equals the macro F1. Any model achieving 96.67% accuracy also achieves a ~96.6% F1-score, confirming SVC and KNN as the top performers.

### Question 4 — What insights can be drawn from the confusion matrices?

```python
print('Per-model error summary:')
for name, y_pred in predictions.items():
    cm     = confusion_matrix(y_test, y_pred)
    errors = len(y_test) - np.trace(cm)
    print(f'  {name:<22} | Errors: {errors} | Class diagonals: {np.diag(cm)}')
```

**Answer — Key insights from the confusion matrices:**

1. **Setosa is perfectly classified by ALL five models.**  
   Every model correctly identifies all 10 Setosa test samples with no confusion. This is because Setosa is linearly separable from the other two species.

2. **All classification errors occur only between Versicolour and Virginica.**  
   These two species overlap in feature space, making them the hardest to distinguish regardless of which algorithm is used.

3. **Lower-accuracy models (Logistic Regression, Decision Tree, Random Forest) make 2 errors** — both in the Versicolour/Virginica boundary zone.

4. **Higher-accuracy models (SVC, KNN) make only 1 error**, still between Versicolour and Virginica.

5. This pattern shows that the limiting factor is the **data itself** (overlapping classes), not the algorithm — no model achieves 100% accuracy on this test set.

### Question 5 — Which model would you recommend and why?

```python
print('Final model scorecard:')
print(results_df[['Model','Accuracy','Precision','Recall','F1-Score']]
      .sort_values('F1-Score', ascending=False).to_string(index=False))
```

**Answer — Recommended Model: Support Vector Classifier (SVC)**

| Criterion | Result |
|-----------|--------|
| Accuracy | 96.67% — tied highest |
| Precision | Highest / tied |
| Recall | Highest / tied |
| F1-Score | ~96.66% — tied highest |
| Precision-Recall balance | Best (near-zero gap) |

**Justification:**  
SVC with an RBF kernel finds the maximum-margin hyperplane between classes, making it especially effective at handling the overlap between Versicolour and Virginica. It achieves top performance as a single model without needing an ensemble.

While **KNN** matches SVC on every metric, SVC is preferred because:
- KNN stores the entire training set and gets slow on larger datasets  
- SVC generalises better through its margin-maximisation principle  
- SVC is more robust to noise in the feature space  

**If interpretability is the priority**, the **Decision Tree** would be recommended instead — its rules can be visualised and explained to a non-technical audience, despite its slightly lower accuracy (93.33%).
